### if문

```python
'''
if 조건문 == True:
  수행할 일
else:
  수행할 일
'''
```

```python
name = '엘사'

if name == '엘사':
    print('찾았습니다!')
else:
    print('계속 찾으세요!')
```

```python
num_members = 2

if num_members == 2:    # 만약 대화방 사용자 수가 2명 이상이면
    print('개인 카톡방입니다')
elif num_members >= 2:
    print('단체 카톡방입니다')
else:
    print('나와의 대화 채팅방입니다')
```

### for문

```python
'''
for 변수 in 자료구조모음:
  수행할 일
'''
```

```python
frozen_group = ['엘사', '안나', '올라프', '크리스토프', '한스']

for name in frozen_group:  # frozen_group 리스트에 있는 이름을 하나씩 출력하세요.
    print(name)
```

```python
for number in range(10): #
    result = number * 2
    print(number, result)
```

## 11.그룹화
- 동일한 값을 가진 것들끼리 합쳐서 통계 또는 평균 등의 값을 계산하기 위해 사용

```python
import pandas as pd
df = pd.read_excel('onepiece.xlsx', index_col=0)
df.index.name = "입단순서"
df
```

```python
df.groupby('고향').get_group('Grandline') # 고향이 Grandline인 사람
```

```python
df.groupby('고향').mean(numeric_only=True) # 계산 가능한 데이터들의 평균값
```

```python
df.groupby('고향').size() # 각 그룹의 크기
```

```python
df.groupby('고향')[['이름','기술']].count() # 고향으로 그룹화를 한 뒤에 고향별 이름,기술 데이터의 수를 가져옴
```
